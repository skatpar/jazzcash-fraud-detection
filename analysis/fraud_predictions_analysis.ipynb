# Fraud Predictions Analysis

This notebook analyzes the fraud predictions generated by the inference script. We'll load the Parquet file containing predictions and explore the results to understand fraud patterns and model performance.

**File to analyze:** `/root/research-dir/dev/jazzcash-fraud-detection/scripts/fraud_predictions_20251106_164517.parquet`

## 1. Import Required Libraries

Import all necessary libraries for data analysis and visualization.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings

# Configure plotting
plt.style.use('default')
sns.set_palette("husl")
warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 50)

print("✅ Libraries imported successfully!")
print(f"📊 Pandas version: {pd.__version__}")
print(f"🔢 NumPy version: {np.__version__}")
print(f"📈 Matplotlib version: {plt.matplotlib.__version__}")
print(f"🎨 Seaborn version: {sns.__version__}")
```

## 2. Load Parquet File

Load the fraud predictions from the Parquet file generated by the inference script.

```python
# Define the path to the Parquet file
parquet_file_path = "/root/research-dir/dev/jazzcash-fraud-detection/scripts/fraud_predictions_20251106_173441.parquet"

# Load the predictions from Parquet file
print("📁 Loading fraud predictions from Parquet file...")
print(f"📍 File path: {parquet_file_path}")

try:
    # Load the Parquet file
    df_predictions = pd.read_parquet(parquet_file_path)
    
    print("✅ Parquet file loaded successfully!")
    print(f"📊 Dataset shape: {df_predictions.shape}")
    print(f"📅 Loaded at: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
except FileNotFoundError:
    print("❌ Error: Parquet file not found!")
    print("Please check if the file exists at the specified path.")
except Exception as e:
    print(f"❌ Error loading file: {str(e)}")
    print("Please check the file format and accessibility.")
```
