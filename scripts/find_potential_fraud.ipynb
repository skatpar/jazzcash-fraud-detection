```python
# Use clickhouse-connect to load data into a pandas DataFrame

import clickhouse_connect
import pandas as pd

# ClickHouse connection parameters
CLICKHOUSE_CONFIG = {
    'host': 'localhost',
    'port': 8123 ,  # HTTP port for clickhouse-connect
    'database': 'public',
    'user': 'default',
    'password': 'DfsTeChB1'
}

# Create ClickHouse client
client = clickhouse_connect.get_client(
    host=CLICKHOUSE_CONFIG['host'],
    port=CLICKHOUSE_CONFIG['port'],
    username=CLICKHOUSE_CONFIG['user'],
    password=CLICKHOUSE_CONFIG['password'],
    database=CLICKHOUSE_CONFIG['database']
)

# Define the query
query = """
    SELECT *
    FROM public.stixor_fraud_features_distributed 
    where cutoff_date between '2025-06-01' and '2025-06-30'
"""

# Execute query and load into pandas DataFrame
df = client.query_df(query)

print(f"✅ Loaded {len(df):,} rows into pandas DataFrame")
df.head()
```
