```python
-- Comprehensive User-Level Feature Engineering Script for PostgreSQL
-- Cut-off Date: 2025-05-31
-- Time Windows: 1d, 3d, 5d, 10d, 20d, 30d

-- Create comprehensive user features table
CREATE TABLE IF NOT EXISTS public.user_comprehensive_features (
    ac_from VARCHAR(50) PRIMARY KEY,
    cutoff_date DATE DEFAULT '2025-05-31',
    
    -- Basic aggregations for different time windows
    -- 1 day features
    total_txns_1d INTEGER DEFAULT 0,
    total_amount_1d DECIMAL(18,2) DEFAULT 0,
    avg_amount_1d DECIMAL(18,2) DEFAULT 0,
    median_amount_1d DECIMAL(18,2) DEFAULT 0,
    max_amount_1d DECIMAL(18,2) DEFAULT 0,
    min_amount_1d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_1d INTEGER DEFAULT 0,
    unique_channels_1d INTEGER DEFAULT 0,
    unique_types_1d INTEGER DEFAULT 0,
    
    -- 3 day features  
    total_txns_3d INTEGER DEFAULT 0,
    total_amount_3d DECIMAL(18,2) DEFAULT 0,
    avg_amount_3d DECIMAL(18,2) DEFAULT 0,
    median_amount_3d DECIMAL(18,2) DEFAULT 0,
    max_amount_3d DECIMAL(18,2) DEFAULT 0,
    min_amount_3d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_3d INTEGER DEFAULT 0,
    unique_channels_3d INTEGER DEFAULT 0,
    unique_types_3d INTEGER DEFAULT 0,
    
    -- 5 day features
    total_txns_5d INTEGER DEFAULT 0,
    total_amount_5d DECIMAL(18,2) DEFAULT 0,
    avg_amount_5d DECIMAL(18,2) DEFAULT 0,
    median_amount_5d DECIMAL(18,2) DEFAULT 0,
    max_amount_5d DECIMAL(18,2) DEFAULT 0,
    min_amount_5d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_5d INTEGER DEFAULT 0,
    unique_channels_5d INTEGER DEFAULT 0,
    unique_types_5d INTEGER DEFAULT 0,
    
    -- 10 day features
    total_txns_10d INTEGER DEFAULT 0,
    total_amount_10d DECIMAL(18,2) DEFAULT 0,
    avg_amount_10d DECIMAL(18,2) DEFAULT 0,
    median_amount_10d DECIMAL(18,2) DEFAULT 0,
    max_amount_10d DECIMAL(18,2) DEFAULT 0,
    min_amount_10d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_10d INTEGER DEFAULT 0,
    unique_channels_10d INTEGER DEFAULT 0,
    unique_types_10d INTEGER DEFAULT 0,
    
    -- 20 day features
    total_txns_20d INTEGER DEFAULT 0,
    total_amount_20d DECIMAL(18,2) DEFAULT 0,
    avg_amount_20d DECIMAL(18,2) DEFAULT 0,
    median_amount_20d DECIMAL(18,2) DEFAULT 0,
    max_amount_20d DECIMAL(18,2) DEFAULT 0,
    min_amount_20d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_20d INTEGER DEFAULT 0,
    unique_channels_20d INTEGER DEFAULT 0,
    unique_types_20d INTEGER DEFAULT 0,
    
    -- 30 day features
    total_txns_30d INTEGER DEFAULT 0,
    total_amount_30d DECIMAL(18,2) DEFAULT 0,
    avg_amount_30d DECIMAL(18,2) DEFAULT 0,
    median_amount_30d DECIMAL(18,2) DEFAULT 0,
    max_amount_30d DECIMAL(18,2) DEFAULT 0,
    min_amount_30d DECIMAL(18,2) DEFAULT 0,
    unique_recipients_30d INTEGER DEFAULT 0,
    unique_channels_30d INTEGER DEFAULT 0,
    unique_types_30d INTEGER DEFAULT 0,
    
    -- Channel preference features
    most_used_channel_30d VARCHAR(50),
    last_used_channel VARCHAR(50),
    channel_diversity_score_30d DECIMAL(5,4) DEFAULT 0,
    
    -- Transaction type features
    most_used_type_30d VARCHAR(50),
    last_used_type VARCHAR(50),
    type_diversity_score_30d DECIMAL(5,4) DEFAULT 0,
    
    -- Time-based features
    night_txns_30d INTEGER DEFAULT 0,
    weekend_txns_30d INTEGER DEFAULT 0,
    peak_hour_txns_30d INTEGER DEFAULT 0,
    off_peak_hour_txns_30d INTEGER DEFAULT 0,
    
    -- Balance features
    avg_start_balance_30d DECIMAL(18,2) DEFAULT 0,
    avg_end_balance_30d DECIMAL(18,2) DEFAULT 0,
    min_balance_30d DECIMAL(18,2) DEFAULT 0,
    max_balance_30d DECIMAL(18,2) DEFAULT 0,
    balance_volatility_30d DECIMAL(18,2) DEFAULT 0,
    
    -- Recipient features
    top_recipient_30d VARCHAR(50),
    avg_amount_per_recipient_30d DECIMAL(18,2) DEFAULT 0,
    max_amount_to_single_recipient_30d DECIMAL(18,2) DEFAULT 0,
    recipient_concentration_ratio_30d DECIMAL(5,4) DEFAULT 0,
    
    -- Behavioral features
    avg_time_between_txns_30d DECIMAL(10,2) DEFAULT 0,
    txn_frequency_score_30d DECIMAL(5,4) DEFAULT 0,
    first_txn_time TIMESTAMP,
    last_txn_time TIMESTAMP,
    days_since_first_txn INTEGER DEFAULT 0,
    days_since_last_txn INTEGER DEFAULT 0,
    
    -- Risk indicators
    large_amount_txns_30d INTEGER DEFAULT 0, -- > 95th percentile
    unusual_time_txns_30d INTEGER DEFAULT 0, -- 2-6 AM
    rapid_succession_txns_30d INTEGER DEFAULT 0, -- < 1 minute apart
    
    processing_timestamp TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Create indexes for better performance
CREATE INDEX IF NOT EXISTS idx_user_features_cutoff_date ON public.user_comprehensive_features(cutoff_date);
CREATE INDEX IF NOT EXISTS idx_user_features_total_txns_30d ON public.user_comprehensive_features(total_txns_30d);
CREATE INDEX IF NOT EXISTS idx_user_features_total_amount_30d ON public.user_comprehensive_features(total_amount_30d);
CREATE INDEX IF NOT EXISTS idx_user_features_processing_timestamp ON public.user_comprehensive_features(processing_timestamp);
```

```python
-- Main feature engineering query with multiple time windows
WITH 
-- Base data with time window filters
user_transactions AS (
    SELECT 
        ac_from,
        ac_to,
        trans_id,
        start_balance,
        end_balance,
        trx_channel,
        trx_type,
        trans_initiate_time,
        data_date,
        CASE 
            WHEN data_date > '2025-05-31' - INTERVAL '1 day' THEN '1d'
            WHEN data_date > '2025-05-31' - INTERVAL '3 days' THEN '3d'
            WHEN data_date > '2025-05-31' - INTERVAL '5 days' THEN '5d'
            WHEN data_date > '2025-05-31' - INTERVAL '10 days' THEN '10d'
            WHEN data_date > '2025-05-31' - INTERVAL '20 days' THEN '20d'
            WHEN data_date > '2025-05-31' - INTERVAL '30 days' THEN '30d'
            ELSE 'older'
        END as time_window,
        EXTRACT(HOUR FROM trans_initiate_time) as txn_hour,
        EXTRACT(DOW FROM trans_initiate_time) as day_of_week
    FROM public.stixor_iar
    WHERE data_date > '2025-05-31' - INTERVAL '30 days'
      AND data_date <= '2025-05-31'
      AND ac_from IS NOT NULL
),

-- Calculate percentiles for risk indicators
amount_percentiles AS (
    SELECT 
        PERCENTILE_CONT(0.95) WITHIN GROUP (ORDER BY start_balance) as p95_amount
    FROM user_transactions
),

-- 1-day features
features_1d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_1d,
        COALESCE(SUM(start_balance), 0) as total_amount_1d,
        COALESCE(AVG(start_balance), 0) as avg_amount_1d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_1d,
        COALESCE(MAX(start_balance), 0) as max_amount_1d,
        COALESCE(MIN(start_balance), 0) as min_amount_1d,
        COUNT(DISTINCT ac_to) as unique_recipients_1d,
        COUNT(DISTINCT trx_channel) as unique_channels_1d,
        COUNT(DISTINCT trx_type) as unique_types_1d
    FROM user_transactions
    WHERE time_window IN ('1d')
    GROUP BY ac_from
),

-- 3-day features
features_3d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_3d,
        COALESCE(SUM(start_balance), 0) as total_amount_3d,
        COALESCE(AVG(start_balance), 0) as avg_amount_3d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_3d,
        COALESCE(MAX(start_balance), 0) as max_amount_3d,
        COALESCE(MIN(start_balance), 0) as min_amount_3d,
        COUNT(DISTINCT ac_to) as unique_recipients_3d,
        COUNT(DISTINCT trx_channel) as unique_channels_3d,
        COUNT(DISTINCT trx_type) as unique_types_3d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d')
    GROUP BY ac_from
),

-- 5-day features
features_5d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_5d,
        COALESCE(SUM(start_balance), 0) as total_amount_5d,
        COALESCE(AVG(start_balance), 0) as avg_amount_5d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_5d,
        COALESCE(MAX(start_balance), 0) as max_amount_5d,
        COALESCE(MIN(start_balance), 0) as min_amount_5d,
        COUNT(DISTINCT ac_to) as unique_recipients_5d,
        COUNT(DISTINCT trx_channel) as unique_channels_5d,
        COUNT(DISTINCT trx_type) as unique_types_5d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d')
    GROUP BY ac_from
),

-- 10-day features
features_10d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_10d,
        COALESCE(SUM(start_balance), 0) as total_amount_10d,
        COALESCE(AVG(start_balance), 0) as avg_amount_10d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_10d,
        COALESCE(MAX(start_balance), 0) as max_amount_10d,
        COALESCE(MIN(start_balance), 0) as min_amount_10d,
        COUNT(DISTINCT ac_to) as unique_recipients_10d,
        COUNT(DISTINCT trx_channel) as unique_channels_10d,
        COUNT(DISTINCT trx_type) as unique_types_10d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d')
    GROUP BY ac_from
),

-- 20-day features
features_20d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_20d,
        COALESCE(SUM(start_balance), 0) as total_amount_20d,
        COALESCE(AVG(start_balance), 0) as avg_amount_20d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_20d,
        COALESCE(MAX(start_balance), 0) as max_amount_20d,
        COALESCE(MIN(start_balance), 0) as min_amount_20d,
        COUNT(DISTINCT ac_to) as unique_recipients_20d,
        COUNT(DISTINCT trx_channel) as unique_channels_20d,
        COUNT(DISTINCT trx_type) as unique_types_20d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d')
    GROUP BY ac_from
),

-- 30-day features
features_30d AS (
    SELECT 
        ac_from,
        COUNT(trans_id) as total_txns_30d,
        COALESCE(SUM(start_balance), 0) as total_amount_30d,
        COALESCE(AVG(start_balance), 0) as avg_amount_30d,
        COALESCE(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY start_balance), 0) as median_amount_30d,
        COALESCE(MAX(start_balance), 0) as max_amount_30d,
        COALESCE(MIN(start_balance), 0) as min_amount_30d,
        COUNT(DISTINCT ac_to) as unique_recipients_30d,
        COUNT(DISTINCT trx_channel) as unique_channels_30d,
        COUNT(DISTINCT trx_type) as unique_types_30d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
    GROUP BY ac_from
)

-- Continue in next cell for channel and behavioral features...
```

```python
-- Channel and behavioral features for 30-day window
channel_features AS (
    SELECT 
        ac_from,
        -- Most used channel
        (array_agg(trx_channel ORDER BY channel_count DESC))[1] as most_used_channel_30d,
        -- Channel diversity (entropy-like measure)
        CASE 
            WHEN COUNT(DISTINCT trx_channel) > 1 
            THEN 1.0 - (MAX(channel_count) * 1.0 / COUNT(*))
            ELSE 0.0 
        END as channel_diversity_score_30d
    FROM (
        SELECT 
            ac_from,
            trx_channel,
            COUNT(*) as channel_count
        FROM user_transactions
        WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
        GROUP BY ac_from, trx_channel
    ) channel_stats
    GROUP BY ac_from
),

-- Transaction type features
type_features AS (
    SELECT 
        ac_from,
        -- Most used type
        (array_agg(trx_type ORDER BY type_count DESC))[1] as most_used_type_30d,
        -- Type diversity
        CASE 
            WHEN COUNT(DISTINCT trx_type) > 1 
            THEN 1.0 - (MAX(type_count) * 1.0 / COUNT(*))
            ELSE 0.0 
        END as type_diversity_score_30d
    FROM (
        SELECT 
            ac_from,
            trx_type,
            COUNT(*) as type_count
        FROM user_transactions
        WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
        GROUP BY ac_from, trx_type
    ) type_stats
    GROUP BY ac_from
),

-- Latest usage features
latest_usage AS (
    SELECT DISTINCT
        ac_from,
        FIRST_VALUE(trx_channel) OVER (PARTITION BY ac_from ORDER BY trans_initiate_time DESC) as last_used_channel,
        FIRST_VALUE(trx_type) OVER (PARTITION BY ac_from ORDER BY trans_initiate_time DESC) as last_used_type
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
),

-- Time-based behavioral features
time_features AS (
    SELECT 
        ac_from,
        -- Night transactions (2-6 AM)
        SUM(CASE WHEN txn_hour BETWEEN 2 AND 6 THEN 1 ELSE 0 END) as night_txns_30d,
        -- Weekend transactions (Saturday=6, Sunday=0)
        SUM(CASE WHEN day_of_week IN (0, 6) THEN 1 ELSE 0 END) as weekend_txns_30d,
        -- Peak hours (9-17)
        SUM(CASE WHEN txn_hour BETWEEN 9 AND 17 THEN 1 ELSE 0 END) as peak_hour_txns_30d,
        -- Off-peak hours
        SUM(CASE WHEN txn_hour NOT BETWEEN 9 AND 17 THEN 1 ELSE 0 END) as off_peak_hour_txns_30d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
    GROUP BY ac_from
),

-- Balance features
balance_features AS (
    SELECT 
        ac_from,
        AVG(start_balance) as avg_start_balance_30d,
        AVG(end_balance) as avg_end_balance_30d,
        MIN(LEAST(start_balance, end_balance)) as min_balance_30d,
        MAX(GREATEST(start_balance, end_balance)) as max_balance_30d,
        STDDEV(start_balance) as balance_volatility_30d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
    GROUP BY ac_from
),

-- Recipient features
recipient_features AS (
    SELECT 
        ac_from,
        -- Top recipient by transaction count
        (array_agg(ac_to ORDER BY recipient_count DESC))[1] as top_recipient_30d,
        -- Average amount per recipient
        AVG(total_amount_to_recipient) as avg_amount_per_recipient_30d,
        -- Max amount to single recipient
        MAX(total_amount_to_recipient) as max_amount_to_single_recipient_30d,
        -- Recipient concentration ratio (top recipient share)
        CASE 
            WHEN SUM(recipient_count) > 0 
            THEN MAX(recipient_count) * 1.0 / SUM(recipient_count)
            ELSE 0.0 
        END as recipient_concentration_ratio_30d
    FROM (
        SELECT 
            ac_from,
            ac_to,
            COUNT(*) as recipient_count,
            SUM(start_balance) as total_amount_to_recipient
        FROM user_transactions
        WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
          AND ac_to IS NOT NULL
        GROUP BY ac_from, ac_to
    ) recipient_stats
    GROUP BY ac_from
),

-- Behavioral timing features
behavioral_features AS (
    SELECT 
        ac_from,
        MIN(trans_initiate_time) as first_txn_time,
        MAX(trans_initiate_time) as last_txn_time,
        DATE_PART('day', '2025-05-31'::date - MIN(data_date)) as days_since_first_txn,
        DATE_PART('day', '2025-05-31'::date - MAX(data_date)) as days_since_last_txn,
        -- Average time between transactions (in hours)
        CASE 
            WHEN COUNT(*) > 1 
            THEN EXTRACT(EPOCH FROM (MAX(trans_initiate_time) - MIN(trans_initiate_time))) / 3600.0 / (COUNT(*) - 1)
            ELSE 0 
        END as avg_time_between_txns_30d,
        -- Transaction frequency score (transactions per day)
        COUNT(*) * 1.0 / GREATEST(DATE_PART('day', MAX(data_date) - MIN(data_date)) + 1, 1) as txn_frequency_score_30d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
    GROUP BY ac_from
),

-- Risk indicators
risk_features AS (
    SELECT 
        ac_from,
        -- Large amount transactions (> 95th percentile)
        SUM(CASE WHEN start_balance > (SELECT p95_amount FROM amount_percentiles) THEN 1 ELSE 0 END) as large_amount_txns_30d,
        -- Unusual time transactions (2-6 AM)
        SUM(CASE WHEN txn_hour BETWEEN 2 AND 6 THEN 1 ELSE 0 END) as unusual_time_txns_30d,
        -- Rapid succession transactions (< 1 minute apart)
        SUM(CASE 
            WHEN LAG(trans_initiate_time) OVER (PARTITION BY ac_from ORDER BY trans_initiate_time) IS NOT NULL
                 AND EXTRACT(EPOCH FROM (trans_initiate_time - LAG(trans_initiate_time) OVER (PARTITION BY ac_from ORDER BY trans_initiate_time))) < 60
            THEN 1 ELSE 0 
        END) as rapid_succession_txns_30d
    FROM user_transactions
    WHERE time_window IN ('1d', '3d', '5d', '10d', '20d', '30d')
    GROUP BY ac_from
)
```

# Comprehensive User-Level Feature Engineering for Fraud Detection

## Overview
This script creates a comprehensive set of user-level features across multiple time windows (1d, 3d, 5d, 10d, 20d, 30d) with cutoff date of **2025-05-31**.

## Features Generated

### Time Window Features (per window)
- **Transaction Counts**: Total transactions, unique channels, unique types
- **Amount Statistics**: Sum, average, median, min, max amounts
- **Recipient Analysis**: Unique recipients count

### Behavioral Features (30-day)
- **Channel Preferences**: Most used channel, last used channel, channel diversity
- **Transaction Types**: Most used type, last used type, type diversity  
- **Temporal Patterns**: Night transactions, weekend transactions, peak/off-peak hours
- **Balance Behavior**: Average start/end balance, min/max balance, volatility

### Recipient Features (30-day)
- **Concentration**: Top recipient, concentration ratio
- **Amount Patterns**: Average amount per recipient, max amount to single recipient

### Risk Indicators (30-day)
- **Large Amounts**: Transactions > 95th percentile
- **Unusual Timing**: Transactions between 2-6 AM
- **Rapid Succession**: Transactions < 1 minute apart

### Time-based Features
- **Activity Timeline**: First/last transaction times, days since first/last
- **Frequency**: Average time between transactions, transaction frequency score

## Key Benefits
1. **Multi-granular Analysis**: Features at 6 different time windows
2. **Behavioral Profiling**: Channel/type preferences and diversity scores
3. **Risk Detection**: Built-in risk indicators for fraud detection
4. **Recipient Analysis**: Understanding money flow patterns
5. **Temporal Intelligence**: Time-based behavioral patterns
