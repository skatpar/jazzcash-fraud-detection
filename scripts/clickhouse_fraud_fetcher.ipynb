# ClickHouse Fraud Data Fetcher

Fetch fraud detection data from ClickHouse into pandas DataFrame.

## Overview
- Connect to ClickHouse database
- Execute SQL query with CTE and GLOBAL JOINs
- Load results into pandas DataFrame
- Display data summary and statistics
- Save to CSV for further analysis

````python
# Import required libraries
import pandas as pd
from clickhouse_driver import Client
import logging
from typing import Optional
from datetime import datetime

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)

print("✅ Libraries imported successfully")
````

````python
# ClickHouse connection configuration
CLICKHOUSE_CONFIG = {
    'host': 'localhost',
    'port': 9000,
    'database': 'public',
    'user': 'default',
    'password': 'DfsTeChB1'
}

print("✅ ClickHouse configuration loaded")
print(f"   • Host: {CLICKHOUSE_CONFIG['host']}")
print(f"   • Port: {CLICKHOUSE_CONFIG['port']}")
print(f"   • Database: {CLICKHOUSE_CONFIG['database']}")
````

````python
class ClickHouseDataFetcher:
    """Fetch fraud detection data from ClickHouse."""
    
    def __init__(self, config: dict):
        """Initialize with ClickHouse config."""
        self.config = config
        self.client = None
    
    def connect(self) -> None:
        """Establish connection to ClickHouse."""
        try:
            self.client = Client(
                host=self.config['host'],
                port=self.config['port'],
                database=self.config['database'],
                user=self.config['user'],
                password=self.config['password']
            )
            logger.info(f"✅ Connected to ClickHouse at {self.config['host']}:{self.config['port']}")
        except Exception as e:
            logger.error(f"❌ Failed to connect: {str(e)}")
            raise
    
    def disconnect(self) -> None:
        """Close connection to ClickHouse."""
        if self.client:
            self.client.disconnect()
            logger.info("🔌 Disconnected from ClickHouse")
    
    def fetch_fraud_data(self) -> pd.DataFrame:
        """Fetch fraud detection data using the provided SQL query."""
        
        # SQL query: Identify frauds and join with transaction data
        query = """ select * from public.fraud_distributed;"""
        
        try:
            logger.info("📊 Fetching fraud detection data from ClickHouse...")
            start_time = datetime.now()
            
            # Fetch data
            data = self.client.execute(query)
            
            df = pd.DataFrame(data)
            
            duration = (datetime.now() - start_time).total_seconds()
            
            logger.info(f"✅ Data fetched successfully in {duration:.2f} seconds!")
            logger.info(f"   • Rows: {len(df):,}")
            logger.info(f"   • Columns: {len(df.columns)}")
            
            return df
        
        except Exception as e:
            logger.error(f"❌ Error fetching data: {str(e)}")
            raise

print("✅ ClickHouseDataFetcher class defined")
````

````python
# Initialize fetcher and connect to ClickHouse
fetcher = ClickHouseDataFetcher(CLICKHOUSE_CONFIG)
fetcher.connect()
````

````python
# Fetch fraud data into pandas DataFrame
df = fetcher.fetch_fraud_data()
````

````python
# Display data information
print("\n" + "="*80)
print("📊 DATA SUMMARY")
print("="*80)

print(f"\n🔍 Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")

print(f"\n📋 Column Names and Types:")
for col, dtype in df.dtypes.items():
    print(f"   • {col:40s} : {dtype}")

print(f"\n📈 Fraud Distribution:")
fraud_counts = df['is_fraud'].value_counts().sort_index()
for label, count in fraud_counts.items():
    label_name = "Fraud" if label == 1 else "Legitimate"
    pct = 100 * count / len(df)
    print(f"   • {label_name:20s} : {count:>10,} ({pct:>6.2f}%)")

print(f"\n❓ Missing Values:")
missing = df.isnull().sum()
if missing.sum() > 0:
    for col, count in missing[missing > 0].items():
        pct = 100 * count / len(df)
        print(f"   • {col:40s} : {count:>10,} ({pct:>6.2f}%)")
else:
    print("   ✅ No missing values found")

print(f"\n📊 Basic Statistics:")
print(df.describe())
````

````python
# Cleanup: disconnect from ClickHouse
fetcher.disconnect()
print("✅ ClickHouse connection closed")
````
