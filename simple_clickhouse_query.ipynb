```python
# Import required libraries
import pandas as pd
from clickhouse_driver import Client
```

## Connect to ClickHouse

```python
# Initialize ClickHouse client
client = Client(
    host='localhost',
    port=9000,
    user='default',
    password='DfsTeChB1',
    database='public'
)

print("✅ Connected to ClickHouse")
```

## Simple Query: Count Records

```python
# Count total records
query = """
 select toMonth(transaction_datetime), trx_type, count(1) as cnt from public.fraud_distributed group by toMonth(transaction_datetime), trx_type order by 1,2;"""

result = client.execute(query)
```

```python
# Convert query result to DataFrame
result_df = pd.DataFrame(result, columns=['month', 'trx_type', 'cnt'])
print(result_df.head())
```
